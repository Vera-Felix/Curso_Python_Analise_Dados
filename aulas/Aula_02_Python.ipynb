
## Operadores aritméticos — Divisão

### 1. Divisão comum (`/`)

Retorna o resultado da divisão, incluindo a parte decimal.

```python
print(10 / 3)  # 3.3333333333333335
```

### 2. Divisão inteira (`//`)

Retorna o quociente arredondado para baixo.

É útil para descobrir quantos grupos completos conseguimos formar.

```python
print(17 // 5)  # 3 grupos completos
```

### 3. Módulo (`%`)

Retorna o resto da divisão.

É útil para descobrir quantos elementos sobraram após formar os grupos completos.

```python
print(17 % 5)  # Sobram 2
```

````python
a = 10
b = 3

print(a / b)   # 3.3333333333333335
print(a // b)  # 3
````

````python
alunas = 17
grupo = 5

print(alunas % grupo)
````

````python
# EXERCÍCIO 1 — Divisão inteira e módulo

# Temos 19 participantes e queremos formar grupos de 6 pessoas.

# 1. Crie uma variável para armazenar o número de participantes.
participantes = 19

# 2. Crie uma variável para armazenar a quantidade de pessoas por grupo.
pessoas_por_grupo = 6

# 3. Calcule quantos grupos completos podem ser formados usando //.
grupos_completos = participantes // pessoas_por_grupo
# 4. Calcule quantas pessoas sobram usando %.
pessoas_sobrando = participantes % pessoas_por_grupo

# 5. Utilize print() para exibir os dois resultados.
print(f"Grupos Completos: {grupos_completos}")
print(f"Pessoas Sobrando: {pessoas_sobrando}")
````


## F-strings — Formatação de textos

As **f-strings** permitem inserir variáveis e expressões diretamente dentro de textos.

Para utilizá-las:
- Colocamos `f` antes das aspas.
- Inserimos as variáveis entre chaves `{}`.
- O Python substitui as expressões entre chaves pelos seus valores.

### Exemplo

```python
nome = "Pretalab"
participantes = 150

print(f"O projeto {nome} possui {participantes} participantes.")
```

**Vantagem:** facilita a construção de mensagens e relatórios com valores dinâmicos.
